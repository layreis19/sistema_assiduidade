```python
from ligacao import conn
import pandas as pd


consulta = """
    SELECT
    id_horario,
    tipo,
    entrada,
    saida,
    inicio_almoco,
    fim_almoco,
    horas_diarias_exigidas,
    CASE
        WHEN tipo = 'FOLGA' THEN 0
        WHEN tipo = 'LIVRE' THEN horas_diarias_exigidas
        WHEN tipo IN ('FIXO', 'TURNO') THEN
            TIME_TO_SEC(TIMEDIFF(saida, entrada)) / 3600
            - TIME_TO_SEC(TIMEDIFF(fim_almoco, inicio_almoco)) / 3600
        ELSE NULL
    END AS total_horas
FROM horario"""

print(consulta)
    
```
